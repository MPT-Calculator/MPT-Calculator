```python
from netgen.occ import *
from netgen.webgui import Draw as DrawGeo
from netgen.meshing import BoundaryLayerParameters
```

```python
# Set materials
material_name = ['Steel','StainSteel']
sigma = [4.5e6, 1.45e6]
mur = [100, 1]# reduced mU-r for testing
alpha =0.01

# Boundary layer options
max_target_frequency = 1e8
number_of_layers = 3
boundary_layer_material = material_name[0]
```

```python

box = Box(Pnt(12,-0.5,-15.75), Pnt(20,0.5,-0.75))

cyl = Cylinder(Pnt(0,0,0), X, r=2.0, h=20)
cyl_inside = Cylinder(Pnt(0,0,0), X, r=1, h=18)
bar_ex = Box(Pnt(0,-2.6,-0.5),(20,-1.6,0.5))
#bar_ex = Box(Pnt(0,-2.5,-1),(20,-1.5,1))

print("Cylinder faces:", len(cyl.faces))
print(cyl.mass)
#cyl=cyl+bar_ex
print("Cylinder faces:", len(cyl.faces))
print(cyl.mass)

#bar = Glue([
#cyl-cyl_inside,
#bar_ex
#])

bar=cyl-cyl_inside
rec=box-cyl
box2 = Box(Pnt(12,-0.8,-3.75), Pnt(20,0.5,-0.75))
rec = rec +box2

bar.bc("default")
rec.bc("default")
#barrel_ex_no_cyl.bc("default")

bar.mat(material_name[0])
#barrel_ex_no_cyl.mat(material_name[0])
rec.mat(material_name[1])

gu = Glue([rec,bar])
```

```python
DrawGeo(gu)
```

```python
# Add large outer box
obox = Box(Pnt(-1000, -1000, -1000), Pnt(1000,1000,1000))
obox.mat('air')
obox.bc('outer')
obox.maxh = 1000
```

```python
# Join gun to the box
obox=obox-gu
joined_object = Glue([gu, obox])
geo = OCCGeometry(joined_object)
#nmesh = geo.GenerateMesh(grading=0.5)

# Setup the boundary layers
mu0 = 4 * 3.14159 * 1e-7
s = alpha / 1e-3
```

```python
# New NG-Solve

if number_of_layers> 0:
    cnt=0
    layers=[]
    for mat in [material_name[0],material_name[1]]:
        print(mat)

        tau = (2/(max_target_frequency * sigma[0] * mu0 * mur[0]))**0.5 /(alpha/s)
        layer_thicknesses = [ (2**n)*tau for n in range(number_of_layers)]
    
        layers.append( BoundaryLayerParameters(boundary=".*", thickness=layer_thicknesses, new_material=mat,
                           domain=mat, outside=False))#, disable_curving=False))
        cnt+=1

    nmesh = OCCGeometry(joined_object).GenerateMesh(meshsize.coarse, boundary_layers=layers)

else:
    nmesh = OCCGeometry(joined_object).GenerateMesh(meshsize.coarse)
```

```python
DrawGeo(nmesh)
```

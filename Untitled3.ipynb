```python
from ngsolve import *
from netgen.occ import *
from netgen.webgui import Draw
```

```python
# 1. Create a reference sphere geometry (radius = 1)
sphere = Sphere((0,0,0), 1.0)
ref_mesh = Mesh(OCCGeometry(sphere).GenerateMesh(maxh=0.1))

# 2. Map the sphere mesh to your high-eccentricity dimensions mathematically
# This guarantees perfectly shaped elements with zero mesher failure!
mapping_cf = CF( (5.02646033 * x, 3.14127911 * y, 0.01624702 * z) )
ref_mesh.SetDeformation(mapping_cf)
Draw(ref_mesh)
```

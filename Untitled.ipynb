```python
import numpy as np
from scipy.optimize import nnls
from scipy.optimize import minimize_scalar


def nnls_tikhonov(A, b, lam, L=None):
    """
    Solve

        min ||Ax-b||^2 + lam ||Lx||^2
        s.t. x >= 0

    using NNLS.
    """
    m, n = A.shape

    if L is None:
        L = np.eye(n)

    A_aug = np.vstack([
        A,
        np.sqrt(lam) * L
    ])

    b_aug = np.concatenate([
        b,
        np.zeros(L.shape[0])
    ])

    x, _ = nnls(A_aug, b_aug)
    return x


def gcv_objective(log_lam, A, b, L=None):
    """
    Approximate GCV score.
    """
    lam = np.exp(log_lam)

    x = nnls_tikhonov(A, b, lam, L)

    residual = np.linalg.norm(A @ x - b) ** 2

    n = A.shape[1]

    if L is None:
        L = np.eye(n)

    ATA = A.T @ A
    LTL = L.T @ L

    H = ATA + lam * LTL

    # Effective degrees of freedom
    df = np.trace(np.linalg.solve(H, ATA))

    denom = (len(b) - df) ** 2

    return residual / max(denom, 1e-12)


def auto_regularized_nnls(A, b, L=None,
                          lambda_min=1e-12,
                          lambda_max=1e3):
    """
    Automatically select lambda via GCV,
    then solve the regularized NNLS problem.
    """

    result = minimize_scalar(
        lambda z: gcv_objective(z, A, b, L),
        bounds=(np.log(lambda_min), np.log(lambda_max)),
        method='bounded'
    )

    lam_opt = np.exp(result.x)

    x = nnls_tikhonov(A, b, lam_opt, L)

    return x, lam_opt


# ------------------------------------------------------------------
# Example
# ------------------------------------------------------------------

if __name__ == "__main__":

    np.random.seed(1)

    m = 10000
    n = 30

    A = np.random.randn(m, n)

    x_true = np.zeros(n)
    x_true[[2, 5, 8, 15]] = [1.0, 2.0, 1.5, 0.8]

    b = A @ x_true + 0.005 * np.random.randn(m)

    # First-order smoothing regularizer
    L = np.eye(n)

    x_est, lam = auto_regularized_nnls(A, b, L)

    print("Selected lambda =", lam)
    print("Solution norm    =", np.linalg.norm(x_est))
    print(x_est)
    print("Residual norm    =", np.linalg.norm(A @ x_est - b))
```

```python
import numpy as np
from scipy.optimize import nnls


def solve_regularized_nnls(A, b, lam, L=None):
    """
    Solve

        min ||Ax - b||² + lam ||Lx||²
        s.t. x >= 0

    via an augmented NNLS system.
    """

    n = A.shape[1]

    if L is None:
        L = np.eye(n)

    A_aug = np.vstack([
        A,
        np.sqrt(lam) * L
    ])

    b_aug = np.concatenate([
        b,
        np.zeros(L.shape[0])
    ])

    x, _ = nnls(A_aug, b_aug)

    return x


def restricted_gcv(A, b, x, tol=1e-10):
    """
    Restricted GCV score.

    df = number of active variables.
    """

    m = len(b)

    residual = np.linalg.norm(A @ x - b) ** 2

    df = np.count_nonzero(x > tol)

    denom = max((m - df) ** 2, 1.0)

    return residual / denom


def auto_regularized_nnls(
        A,
        b,
        L=None,
        lambda_min=1e-8,
        lambda_max=1e2,
        n_lambda=40,
        tol=1e-10):
    """
    Automatic lambda selection using restricted GCV.
    """

    lambdas = np.logspace(
        np.log10(lambda_min),
        np.log10(lambda_max),
        n_lambda
    )

    best_x = None
    best_lambda = None
    best_score = np.inf

    scores = []

    for lam in lambdas:

        x = solve_regularized_nnls(A, b, lam, L)

        score = restricted_gcv(A, b, x, tol)

        scores.append(score)

        if score < best_score:
            best_score = score
            best_lambda = lam
            best_x = x

    return {
        "x": best_x,
        "lambda": best_lambda,
        "gcv": best_score,
        "lambdas": lambdas,
        "scores": np.array(scores)
    }


# ------------------------------------------------------------------
# Example
# ------------------------------------------------------------------

if __name__ == "__main__":

    np.random.seed(123)

    m = 10000
    n = 60

    A = np.random.randn(m, n)

    x_true = np.zeros(n)
    x_true[[5, 12, 20, 31]] = [2.0, 1.5, 3.0, 1.0]

    noise = 0.005 * np.random.randn(m)

    b = A @ x_true + noise

    # Identity regularization
    L = np.eye(n)

    result = auto_regularized_nnls(
        A,
        b,
        L=L,
        lambda_min=1e-8,
        lambda_max=1e1,
        n_lambda=50
    )

    print(f"Selected lambda = {result['lambda']:.4e}")
    print(f"GCV score       = {result['gcv']:.4e}")

    x_est = result["x"]

    print(f"Active variables = {np.count_nonzero(x_est > 1e-10)}")
    print(f"Residual norm    = {np.linalg.norm(A @ x_est - b):.4e}")
```

```python
import webbrowser
from netgen.occ import *
from netgen.webgui import Draw

sp = Sphere( (0,0,0), 1)
r1 = 2.93348328
r2 = 1.90849167
r3 = 1.3700185
gtr = gp_GTrsf( (r1,0,0, 0,r2,0, 0,0,r3), (0,0,0))
ell = gtr (sp)
#ell =sp
Draw (ell)
```

```python
pos_ellipsoid = ell - Box(Pnt(-100,-100,-100), Pnt(0,100,100))
neg_ellipsoid = ell - Box(Pnt(0,-100,-100), Pnt(100,100,100))
ell2= pos_ellipsoid + neg_ellipsoid
```

```python
Draw (ell)
```

```python
ell.maxh = 0.3
ell.bc('default')
ell.mat('object')
```

```python
box = Box(Pnt(-1000, -1000, -1000), Pnt(1000,1000,1000))
box.mat('air')
box.bc('outer')
box.maxh=1000
box=box-ell
```

```python
joined_object = Glue([ell, box])
```

```python
max_target_frequency = 1e8
boundary_layer_material = 'object'
number_of_layers = 2
sigma=[1e7]
mur=[40]
alpha=1e-3

mu0 = 4 * 3.14159 * 1e-7
tau = (2/(max_target_frequency * sigma[0] * mu0 * mur[0]))**0.5 / alpha
layer_thicknesses = [(2**n)*tau for n in range(number_of_layers)]
```

```python
from netgen.meshing import BoundaryLayerParameters
B = BoundaryLayerParameters(boundary=".*", thickness=layer_thicknesses, new_material=boundary_layer_material,
                           domain=boundary_layer_material, outside=False, disable_curving=False )
nmesh = OCCGeometry(joined_object).GenerateMesh(meshsize.coarse,boundary_layers=[B])
nmesh.Refine()
help(nmesh)
```

```python
Draw(nmesh)
from ngsolve import Mesh
mesh=Mesh(nmesh)
mesh.Refine()
mesh.Curve(3)
Draw(mesh)
mesh.GetNE()
```
